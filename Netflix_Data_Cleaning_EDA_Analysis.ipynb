**Addressing Null Values**

```python
netflix['show_id'].nunique()
```

```python
netflix.isnull().sum()
```

```python
netflix['director'].fillna('Unknown', inplace=True)
```

```python
netflix.shape
```

```python
netflix = netflix.dropna(subset=['date_added', 'rating', 'duration'])
```

```python
netflix.shape
```

```python
netflix[['director', 'cast', 'country']].head()
```

```python
netflix[['director', 'cast', 'country']] = netflix[['director', 'cast', 'country']].fillna('Unknown')
```

```python
netflix.isnull().sum()
```

**Standardizing Data**

```python
netflix.info()
```

```python
text_columns = netflix.select_dtypes(include='object').columns
text_columns
```

```python
netflix[text_columns] = netflix[text_columns].replace(
    {r'\r': ' ', r'\n': ' '},
    regex=True
)
```

```python
netflix[text_columns] = netflix[text_columns].apply(lambda x: x.str.strip())
```

```python
netflix['date_added'] = pd.to_datetime(netflix['date_added'], format= 'mixed')
```

```python
print(netflix['type'].unique())
print(netflix['rating'].unique())
print(netflix['duration'].unique())
```

```python
netflix[
    ~netflix['duration'].str.contains('min|Season', case=False, na=False)
]['duration'].unique()
```

```python
netflix['release_year'].min(), netflix['release_year'].max()
```

**Visualization**

```python
type_counts = netflix['type'].value_counts()

type_counts.plot(kind='bar')

plt.title('Distribution of Movies and TV Shows')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')
plt.xticks(rotation=0)
plt.show()
```

The chart shows that movies make up the majority of Netflix content in this dataset, with 6,131 movies compared to 2,676 TV shows.

```python
content_by_year = netflix['year_added'].value_counts().sort_index()

content_by_year.plot(kind='line', marker='o')

plt.title('Number of Netflix Titles Added by Year')
plt.xlabel('Year Added')
plt.ylabel('Number of Titles')
plt.grid(True)
plt.show()
```

The line chart shows that Netflix content additions had a rapid increase from 2016 onward. The highest number of titles was added in 2019, after which the number of added titles declined slightly in 2020 and 2021.

```python
top_countries = netflix['country'].value_counts().head(10)

top_countries.plot(kind='barh')

plt.title('Top 10 Countries by Number of Netflix Titles')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.gca().invert_yaxis()
plt.show()
```

This chart shows that the United States has the highest number of Netflix titles in the dataset, with 2,809 titles. India comes second with 972 titles, followed by the United Kingdom with 418 titles.

```python
rating_counts = netflix['rating'].value_counts()

rating_counts.plot(kind='bar')

plt.title('Distribution of Netflix Content Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

The chart shows that TV-MA is the most common content rating, followed by TV-14. This suggests that a large portion of Netflix titles in the dataset are targeted toward mature and teenage audiences.

```python
top_genres = genres.value_counts().head(10)

top_genres.plot(kind='barh')

plt.title('Top 10 Netflix Genres')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.gca().invert_yaxis()
plt.show()
```

This chart shows that International Movies and Dramas are the most common genres in the Netflix dataset. International Movies appear 2,752 times, while Dramas appear 2,426 times.

**Netflix Dataset Key Insights**

The following key insights were discovered;

*  Content addition sharply increased from 2016 onward, decliing slightly in 2020 and 2021. In the year 2019, Netflix added the highest number of titles.
*   Movies make up the majority of Netflix content in the dataset, with 6,131 movies compared to 2,676 TV shows.
*   International Movies and Dramas are the most common genres, showing that Netflix has a strong collection of international and drama-related content.
*   TV-MA is the most common content rating, followed by TV-14, suggesting that a large portion of Netflix content is targeted toward mature and teenage audiences.
*   The United States has the highest number of Netflix titles, contributing 2,809 titles, followed by India and the United Kingdom.
